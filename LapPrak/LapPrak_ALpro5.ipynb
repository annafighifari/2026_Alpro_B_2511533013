### 2.2.1 Program "1_2511533013"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# jumlah_genap_2511533013.py
ulang_3013 = int(input("Masukkan nilai batas: "))

jumlah_3013 = 0
for i_3013 in range(1, ulang_3013 + 1):
    if i_3013 % 2 == 0:
        print(i_3013, end=" ")
        jumlah_3013 = jumlah_3013 + i_3013

        if i_3013 < ulang_3013:
            print("+", end=" ")
        else:
            print("=", jumlah_3013, end=" ")
print()
print("Jumlah =", jumlah_3013)
```

**Penjelasan :**
1. Perulangan dan Seleksi Kondisi: Program menggunakan struktur loop for i_3013 in range(1, ulang_3013 + 1) yang dikombinasikan dengan seleksi if if i_3013 % 2 == 0:. Kondisi modulo 2 sama dengan 0 ini memastikan bahwa hanya bilangan genap saja yang akan diproses, dicetak, dan ditambahkan ke dalam total penjumlahan, sedangkan bilangan ganjil dilewati.

2. Akumulasi Total: Terdapat variabel jumlah_3013 = 0 yang dideklarasikan di luar perulangan. Setiap kali iterasi menemukan bilangan genap, nilai tersebut ditambahkan ke akumulator melalui operasi jumlah_3013 = jumlah_3013 + i_3013.

3. Format Cetakan Visual: Untuk menghasilkan bentuk penjumlahan matematis yang rapi (seperti "2 + 4 + 6 = 12"), digunakan kondisi bersarang tambahan. Selama nilai genap yang berjalan masih lebih kecil dari batas perulangan (i_3013 < ulang_3013), program akan mencetak karakter + di sebelah angka genap. Jika iterasi telah mencapai batas akhir genap, program mencetak = diikuti hasil total perhitungannya.


```python
#  nested_for1_2511533013.py
batas_3013 = int(input("Masukkan nilai batas: "))
for line_3013 in range(1, batas_3013 + 1):
    for j_3013 in range(1, (-1 * line_3013 + batas_3013) + 1):
        print(".", end=" ")
    print(line_3013)
```

**Penjelasan :**
1. Fungsi Outer Loop: Blok outer loop menggunakan for line_3013 in range(1, batas_3013 + 1) untuk mengatur pergerakan menurun antar baris. Nilai line_3013 ini jugalah yang dicetak pada setiap akhir baris (pindah baris) menggunakan print(line_3013).   
2. Fungsi Inner Loop: Inner loop menangani pencetakan karakter titik secara mendatar tanpa membuat baris baru akibat penggunaan parameter end=" " pada fungsi print(".", end=" ").   
3. Dinamika Pengurangan Titik: Bagian paling krusial adalah rentang perhitungan dalam inner loop: range(1, (-1 * line_3013 + batas_3013) + 1). Karena nilai line_3013 terus bertambah (+1) pada tiap baris, maka hasil pengurangannya terhadap batas_3013 akan menyusut, sehingga jumlah titik yang dicetak terus berkurang dari atas ke bawah membentuk bangun segitiga.


```python
# nested_for2_2511533013.py
batas_3013 = int(input("Masukkan nilai batas: "))
for i_3013 in range(1, batas_3013 + 1):
    for j_3013 in range(1, batas_3013 + 1):
        print("*", end=" ")
    print(i_3013)
```

**Penjelasan :**
1. Struktur Grid (Baris x Kolom): Terdiri dari outer loop i_3013 dan inner loop j_3013, di mana kedua perulangan ini menggunakan rentang range yang persis sama, yaitu dari 1 hingga batas_3013 + 1. Kesamaan rentang ini menghasilkan pola kotak atau matriks berdimensi batas x batas.   
2. Karakter Statis vs Dinamis: Di dalam iterasi horizontal (inner loop), karakter yang dicetak selalu sama (statis), yakni bintang (*). Elemen dinamisnya hanyalah nomor baris dari variabel i_3013 (seperti 1, 2, 3...) yang ditempatkan sesudah iterasi karakter bintang selesai pada baris tersebut.   
3. Manajemen Pemindahan Baris: Penggabungan print("*", end=" ") dan print(i_3013) di level outer loop memastikan karakter bintang dicetak berjajar pada satu baris, dan penulisan angka i_3013 sekaligus memberikan perintah newline atau pindah baris secara implisit sebelum melanjutkan ke siklus baris berikutnya. 


```python
# nested_for3_2511533013.py
batas_3013 = int(input("Masukkan nilai batas: "))
for i_3013 in range(batas_3013 + 1):
    for j_3013 in range(batas_3013 + 1):
        print(i_3013 + j_3013, end=" ")
    print(i_3013)
```

**Penjelasan :**
1. Indeks Berawal Nol: Secara default, struktur range(batas_3013 + 1) yang tidak disebutkan angka awalannya akan memulai perhitungannya secara zero-based indexing (dari angka 0). Jadi jika batas adalah 3, iterasi berputar pada angka 0, 1, 2, dan 3.   
2. Formula Matriks Penjumlahan: Setiap sel angka pada pola ini dicetak dengan operasi matematis dalam fungsi cetak: print(i_3013 + j_3013, end=" "). Ini berarti pada baris ke-0 (i=0) nilainya murni deret kolom (j), pada baris ke-1 (i=1) semua nilai kolom (j) akan bertambah satu, dan seterusnya.   
3. Penutup Baris Numerik: Sama halnya dengan latihan ke-2, baris matriks ini selalu ditutup dengan pencetakan outer variable (i_3013) sebagai identitas baris tersebut, contohnya mencetak print(i_3013) yang akan sekaligus menekan enter ke baris baru. 


```python
# nested_for4_2511533013.py
tinggi_3013 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3013 % 2 != 0:
    print("Tinggi pola harus bilangan genap.")
else:
    a_3013 = tinggi_3013 
    c_3013 = a_3013
    lebar_3013 = (2 * tinggi_3013) - 2

    for i_3013 in range(1, tinggi_3013 + 1):
        b_3013 = c_3013 + 1

        for j_3013 in range(1, lebar_3013 + 1):

            # baris atas dan bawah
            if i_3013 == 1 or i_3013 == tinggi_3013:
                if j_3013 == 1 or j_3013 == lebar_3013:
                    print("#", end="")
                else:
                    print("=", end="")
                    # baris isi
            else:
                if j_3013 == 1 or j_3013 == lebar_3013:
                    print("|", end="")
                else:
                    if j_3013 == c_3013:
                        print("<", end="")
                    elif j_3013 == b_3013:
                        print(">", end="")
                    elif j_3013 == (lebar_3013 - c_3013):
                        print("<", end="")
                    elif j_3013 == (lebar_3013 - c_3013 + 1):
                        print(">", end="")
                    elif j_3013 > b_3013 and j_3013 < (lebar_3013 - c_3013):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()
        # logic asli java
        a_3013 -= 2
        if a_3013<= 0:
            c_3013 = (-a_3013) + 2
        else:
            c_3013 = a_3013  
```

**Penjelasan :**
1. Syarat Genap dan Persiapan Lebar: Program memblokir bilangan ganjil dengan seleksi kondisi if tinggi_3013 % 2 != 0: karena bentuk pola menuntut tinggi yang bisa dibagi dua secara simetris. Ditetapkan juga kalkulasi lebar_3013 = (2 * tinggi_3013) - 2 untuk menemukan ukuran absolut lebar matriks yang dibutuhkan sebagai basis rentang inner loop horizontal.   
2. Pembuatan Dinding (Bingkai Luar): Kondisi pertama mengecek posisi matriks atas dan bawah (i==1 atau i==tinggi), memberikan simbol ujung siku (#) di kolom pertama dan terakhir, dan mengisi bingkai tengah (=) bersambung. Untuk isi badan, pinggiran kiri-kanan ditutup oleh dinding pembatas (|).   
3. Manipulasi Batas Pola Dalam (Inner Shape Logic): Empat buah variabel pengecekan utama seperti c, b, serta pengurang terhadap sisa lebar digunakan dalam chain if-elif kompleks (if j_3013 == c_3013: ... elif j_3013 == b_3013: ...). Posisi kursor kolom memandu di mana karakter buka-kurung <, tutup-kurung >, dan titik . ditempatkan. Operasi manual terhadap variabel batas (seperti a_3013 -= 2 di akhir setiap baris) diimplementasikan layaknya manipulasi iterator pada bahasa C/Java, mendesak titik pertemuan panah agar mengecil dan kembali melebar.

3. Deret Angka Start dari Satu: Pada loop kedua, jangkauan range dimodifikasi dengan penentuan batas bawah dan atas secara spesifik yakni range(1, ulang_3013+1):. Modifikasi pada parameter dasar range ini membuat cetakan angka sekarang dimulai persis dari angka 1 dan berhenti tepat di nilai maksimal input yang dimasukkan.

**Penjelasan :**
1. Deret Penjumlahan Murni: Looping dibentuk dengan parameter for i_3013 in range(1, ulang_3013 + 1): sehingga akan mengeksekusi dan memproses seluruh deret bilangan tanpa ada syarat pembatasan tambahan.   
2. Akumulasi Keseluruhan: Serupa dengan struktur akumulasi umumnya, variabel di luar blok jumlah_3013 = 0 digunakan dan dijumlahkan berulang kali dengan metode jumlah_3013 = jumlah_3013 + i_3013 di mana angka yang diakumulasikan menampung rentetan angka natural (1, 2, 3, 4, dst).   
3. Logika Tanda Operator: Menggunakan percabangan bersarang berupa logika if dan else terhadap variabel kontrol perulangan (i_3013 < ulang_3013). Selama perulangan itu bukanlah iterasi terakhir, maka akan terus mengeluarkan simbol " + ". Namun pada baris akhir putaran (else), akan merangkum dengan mencetak simbol " = " serta nilai total akumulasi deret.

Formula ini berarti baris teratas (saat i bernilai kecil) akan mendapat dorongan spasi paling banyak, dan semakin ke bawah (saat i membesar mendekati nilai tinggi), spasi tersebut akan menyusut hingga nol.   
3. Formula Deret Ganjil Bintang: Pembentukan bentuk segitiga dicapai menggunakan perhitungan pembentuk deret ganjil yaitu bintang_3013 = "*" * (2 * i_3013 - 1). Formula ini menjamin jumlah bintang yang dicetak bertambah dua secara konstan setiap barisnya (1, 3, 5, 7, ...), sehingga membentuk kemiringan sisi piramida secara sempurna ketika digabungkan dengan spasi sebelumnya.

```python
# tugas5_2511533013.py
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_3013 = int(input("Masukkan ukuran skala jam pasir (N): "))

print("#", end="")
for i_3013 in range(4 * n_3013 + 5):
    print("=", end="")
print("#")

for baris_3013 in range(n_3013, 0, -1):
    print("| ", end="")
    
    for spasi_kiri_3013 in range(2 * (n_3013 - baris_3013)):
        print(" ", end="")
        
    for angka_kiri_3013 in range(baris_3013, 0, -1):
        print(f"{angka_kiri_3013} ", end="")
        
    print("<*>", end="")
    
    for angka_kanan_3013 in range(1, baris_3013 + 1):
        print(f" {angka_kanan_3013}", end="")
        
    for spasi_kanan_3013 in range(2 * (n_3013 - baris_3013)):
        print(" ", end="")
        
    print(" |")

print("|", end="")
for spasi_poros_kiri_3013 in range(2 * n_3013 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_poros_kanan_3013 in range(2 * n_3013 + 1):
    print(" ", end="")
print("|")

for baris_3013 in range(1, n_3013 + 1):
    print("| ", end="")
    
    for spasi_kiri_3013 in range(2 * (n_3013 - baris_3013)):
        print(" ", end="")
        
    for angka_kiri_3013 in range(baris_3013, 0, -1):
        print(f"{angka_kiri_3013} ", end="")
        
    print("<*>", end="")
    
    for angka_kanan_3013 in range(1, baris_3013 + 1):
        print(f" {angka_kanan_3013}", end="")
        
    for spasi_kanan_3013 in range(2 * (n_3013 - baris_3013)):
        print(" ", end="")
        
    print(" |")

print("#", end="")
for i_3013 in range(4 * n_3013 + 5):
    print("=", end="")
print("#")
```

**Penjelasan :**
1. Struktur Frame Horizontal Menggunakan Loop: Berbeda dengan teknik manipulasi string = * n, program ini taat murni memakai loop dengan rumus pencetak border range(4 * n_3013 + 5) yang berjalan horizontal mendirikan karakter "=", menyesuaikan proporsi lebar terbesar jam pasir seiring makin besarnya variabel skal
2. Dinamika Loop Bertingkat Fase 1 (Reduksi): Fase awal diolah melalui outer loop decrement for baris_3013 in range(n_3013, 0, -1)
3. Singularity dan Fase 3 (Ekspansi Berbalik): Pada inti pusat (titik poros), jarak pemisah langsung dihitung sebesar (2 * N + 1) di kiri dan kanannya tanpa outer loop. Fase ekspansi (Fase 3 bawah) secara esensial adalah pengulangan sama persis kode-kode pada Fase 1, namun logika pemanggil outer loop direkayasa sebaliknya dari decrement (baris turun ke 1) menjadi blok fungsi increment (dari 1 terus menanjak dan melebar ke ukuran N maksimal).
